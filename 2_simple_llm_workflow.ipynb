```python
from langgraph.graph import StateGraph, START, END 
from langchain_openai import ChatOpenAI 
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
# default model here loaded chatopenAI
model = ChatOpenAI()
```

```python
# creat a State
class LLMState(TypedDict):
    
    question: str
    answer: str
```

```python
def llm_qa(state: LLMState) -> LLMState:
    
    # extract the question from the state
    question = state['question']
    
    # form a prompt for the LLM
    prompt = f"Answer the following question: {question}"
    
    # ask that question to the LLM ( .content beacus llm pass so much thing we only want content attribute beacus inside that we have of answer)
    answer = model.invoke(prompt).content
    
    # update the answer in the state
    state['answer'] = answer
    
    return state
    
```

```python
# create our graph

graph = StateGraph(LLMState)

# add nodes to the graph
graph.add_node('llm_qa', llm_qa)

# add edges to the graph
graph.add_edge(START, 'llm_qa')
graph.add_edge('llm_qa', END)

# compile our graph
workflow =graph.compile()
```

```python
# excute the graph with an initial state
initial_state = {"question": "What is the capital of France?"}

final_state = workflow.invoke(initial_state)

print(final_state)
# only answer print then print(final_state['answer'])
```
